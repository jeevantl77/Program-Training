```python
n=int(input("enter no:of candies available in the jar"))
k=int(input("enter number of minimum candies that must be inside JAR ever."))
C=1
R=n
while(C==1):
  M=int(input("Enter amout of candies needed:"))
  if(M>n):
    print("invalid")
  else:
    R=R-M
    if(R<=k):
      R=n
    print("SOLD=",M)
    print("Remaining=",R)
    C=int(input("DO you want to continue(1or0):"))
```
