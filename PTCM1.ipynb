```python
n=int(input("enter n"))
k=int(input("enter k"))
j=int(input("enter j"))
m=int(input("enter m"))
p=int(input("enter p"))
if k<0 or j<0:
      print("invalid input")

noofmonkeys=m//k+p//j


n=n-noofmonkeys
if m%k!=0 or p%j!=0:
      n=n-1
print(n)6
```
