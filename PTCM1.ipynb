```python
list=[]
n=int(input("Ener list size"))
for i in range(n):
  b=int(input("enter elements"))
  list.append(b)
c=int(input("Enter the element to be found="))
s=0
if c in list:
  for i in list:
    if c==i:
      s=s+1
  print("Found",s,"times")
else:
  print("Not found")
```
